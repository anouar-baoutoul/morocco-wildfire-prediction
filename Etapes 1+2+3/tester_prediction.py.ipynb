```python
import joblib
import pandas as pd
from pathlib import Path

# 1. Charger le modèle et le scaler sauvegardés
OUT = Path('resultats')
model = joblib.load(OUT / 'modele_xgb_optimise.pkl')
scaler = joblib.load(OUT / 'scaler.pkl')

# 2. Simuler de nouvelles données météo (par exemple : une canicule estivale sans pluie)
# Ordre des 8 features : lat_c, lon_c, mois, jour_annee, temperature_2m_max, precipitation_sum, wind_speed_10m_max, humidite_moy
nouvelle_observation = pd.DataFrame([[
    31.6295, -7.9811, 8, 220, 42.5, 0.0, 25.0, 20.0
]], columns=['lat_c', 'lon_c', 'mois', 'jour_annee', 'temperature_2m_max', 'precipitation_sum', 'wind_speed_10m_max', 'humidite_moy'])

# 3. Normaliser les données avec le scaler sauvegardé
observation_scaled = pd.DataFrame(scaler.transform(nouvelle_observation), columns=nouvelle_observation.columns)

# 4. Faire la prédiction et récupérer les probabilités
probabilite = model.predict_proba(observation_scaled)
prob_feu = probabilite[0][1]

# 5. Application d'un seuil d'alerte intelligent (ex: 35%)
seuil_alerte = 0.35

print(f"--- RÉSULTAT DE LA PRÉDICTION ---")
if prob_feu >= seuil_alerte:
    print(f"🚨 ALERTE : Risque d'incendie détecté !")
    print(f"Probabilité de feu : {prob_feu*100:.2f}%")
else:
    print(f"✅ Risque faible ou nul de feu.")
    print(f"Probabilité de feu : {prob_feu*100:.2f}%")
```
